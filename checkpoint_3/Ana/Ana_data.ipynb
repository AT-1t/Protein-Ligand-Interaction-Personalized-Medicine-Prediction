```python
#path genes from bBioPortal 
#l858r_df = data[data["EGFR"].str.contains("L858R", case=False, na=False)]
#print(l858r_df.head())

"""
path genes from cBioPortal from Lung Adenocarcinoma (TCGA, PanCancer Atlas) - EGFR mutations

mutation subtype stratification; point mutations: L858R and exon 19 deletions (need copy numbers (CNV))
"""
#find a way to remove for loop (dock points)
def classify_mutation(mutation):
    if pd.isna(mutation):
        return "Unknown"
    m = mutation.lower()
    if "l858r" in m:
        return "L858R"
    elif "del" in m or "e746" in m or "exon19" in m:
        return "Exon19"
    else:
        return "Other"

egfr_group_counts = mutation['EGFR_group'].value_counts()
print("Frequency of each EGFR group type:")
display(egfr_group_counts)
```
