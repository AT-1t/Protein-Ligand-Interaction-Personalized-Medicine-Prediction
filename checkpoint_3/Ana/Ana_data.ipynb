```python
import pandas as pd
```

```python
data = pd.read_csv('check_EGFR.csv')
data.head()
```

```python
# Check and remove missing values
print("Shape before removing NaN:", data.shape)
print("\nMissing values per column:")
missing = data.isnull().sum()
print(missing[missing > 0])

print("\nColumn names available:")
print(data.columns.tolist())

# Remove rows with missing values in key columns
key_cols = ['Disease-specific Survival status', 'Diagnosis Age', 'Sex']
data_clean = data.dropna(subset=key_cols)
print(f"\nShape after removing NaN: {data_clean.shape}")
print(f"Rows removed: {data.shape[0] - data_clean.shape[0]}")

data_clean.head()
```
